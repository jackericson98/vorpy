```python
import os
os.chdir("../..")
from System.system import System, Edge, Surface
from Visualize.mpl_visualize import plot_atoms, plot_verts, plot_surfs, plot_edges
import matplotlib.pyplot as plt
```

```python
# Create atom objects from sets of points
atoms = [[[2, 0, -1], 5]]
dist = 5
rad = 2

atoms += [[[dist, 0, 2], rad+2], [[0, 0, dist], rad+2], [[0, dist, 0], rad+2]]
sys = System(atoms=atoms)
print(sys.name)
sys.build_network(sol_verts=False, surf_res=0.2, max_vert=5, box_size=1.5)
print(len(sys.net.verts))
```

```python
# Set up the plot
%matplotlib notebook
fig = plt.figure()
ax = fig.add_subplot(projection="3d")
ax.set_title("Basic Cube Cell")
# Plot the elements of the network
# plot_atoms(sys.atoms, fig=fig, ax=ax, dfo=2, alpha=0.1, colors=['w', 'r', 'r', 'r'])
plot_verts(sys.net.verts, plot_spheres=True, fig=fig, ax=ax, Show=True)
plot_edges(sys.net.edges, fig=fig, ax=ax)
plot_surfs(sys.net.surfs, fig=fig, ax=ax, Show=True, dfo=10)
```
